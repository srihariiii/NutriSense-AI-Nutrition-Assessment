# Task 4

```python
## Step 1: Merge NHANES Datasets

df = demo.merge(dr1tot, on="person_id", how="inner")
df = df.merge(cbc, on="person_id", how="left")
df = df.merge(vid, on="person_id", how="left")
df = df.merge(fertin, on="person_id", how="left")

```

```python
## Step 2: Keep Adults (18–65 Years)

df = df[(df["age_years"] >= 18) & (df["age_years"] <= 65)].copy()
```

```python
## Step 3: Preview the Merged Dataset
parquet_path = os.path.join(nhanes_path, "nhanes_merged_adults.parquet")
df.to_parquet(parquet_path, index=False)

print(f"\nParquet file saved successfully.")
print("Location:", parquet_path)

print("Merged Dataset (Adults 18–65)")
print("Shape:", df.shape)

display(df.head())
```

```python
import numpy as np
## Step 3: Remove Rows with Missing Required Columns

COLUMNS_NEEDED = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

df = df.dropna(subset=COLUMNS_NEEDED).copy()

## Step 4: Calculate Iron Label

low_hemoglobin = (
    ((df["gender"] == 1) & (df["hemoglobin_g_dl"] < 13))
    | ((df["gender"] == 2) & (df["hemoglobin_g_dl"] < 12))
)
low_ferritin = df["ferritin"] < 30
lab_iron_risk = low_hemoglobin | low_ferritin
has_iron_lab = df["hemoglobin_g_dl"].notna() | df["ferritin"].notna()
low_iron_intake = df["iron_mg"] < 8
df["label_iron"] = np.where(
    has_iron_lab, lab_iron_risk.fillna(False), low_iron_intake
).astype(int)

## Step 5: Calculate Vitamin D Label

has_vitd_lab = df["serum_vitamin_d"].notna()

lab_vitd_risk = df["serum_vitamin_d"] < 50

low_vitd_intake = df["vitamin_d_mcg"] < 15

df["label_vitamin_d"] = np.where(
    has_vitd_lab,
    lab_vitd_risk,
    low_vitd_intake
).astype(int)

## Intake-only labels
df["label_vitamin_b12"] = (df["vitamin_b12_mcg"] < 2.4).astype(int)
df["label_calcium"] = (df["calcium_mg"] < 1000).astype(int)
df["label_folate"] = (df["folate_mcg"] < 400).astype(int)
df["label_vitamin_a"] = (df["vitamin_a_mcg"] < 700).astype(int)
df["label_vitamin_c"] = (df["vitamin_c_mg"] < 75).astype(int)
df["label_magnesium"] = (df["magnesium_mg"] < 320).astype(int)
df["label_zinc"] = (df["zinc_mg"] < 8).astype(int)
df["label_protein"] = (df["protein_g"] < 50).astype(int)
```

```python
## Step 14: Label Columns

LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]
```

```python
## Step 15: Preview Final Dataset

display(df.head(10))
print("Rows, Columns :", df.shape)
```

```python
## Label Balance

print("Label balance (0 vs 1):")

for col in LABEL_COLS:
    counts = df[col].value_counts().to_dict()
    positive = df[col].mean() * 100
    print(f"{col:22} → {counts} | positive={positive:.1f}%")
```

```python
## Step 1: Import Required Libraries

import time
import joblib
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.multioutput import MultiOutputClassifier

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score

from xgboost import XGBClassifier
```

```python
## Step 2: Prepare Features and Labels

X = df[COLUMNS_NEEDED]

Y = df[LABEL_COLS]

print("Feature Shape :", X.shape)

print("Label Shape :", Y.shape)

## Step 3: Train-Test Split

X_train, X_test, Y_train, Y_test = train_test_split(

    X,

    Y,

    test_size=0.20,

    random_state=42

)

print("Training Samples :", X_train.shape)

print("Testing Samples :", X_test.shape)
```

```python
## Step 3.5: Preprocess Training and Testing Data

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

preprocess = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

X_train_scaled = preprocess.fit_transform(X_train)
X_test_scaled = preprocess.transform(X_test)

print("Training Shape:", X_train_scaled.shape)
print("Testing Shape :", X_test_scaled.shape)
```

```python
## Step 4: Train Random Forest

## Step 5: Train Random Forest Model

import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier

print("Training Random Forest ...")

start = time.time()

rf_model = MultiOutputClassifier(
    RandomForestClassifier(
        n_estimators=200,
        max_depth=14,
        min_samples_leaf=5,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1,
    )
)

rf_model.fit(X_train_scaled, Y_train)

print(f"RF done in {time.time()-start:.1f}s")

rf_pred = rf_model.predict(X_test_scaled)
```

```python
## Step 6: Evaluate Random Forest

from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
import pandas as pd

print("\n" + "="*70)
print("SCORES — Random Forest")
print("="*70)

overall_acc = (Y_test.values == rf_pred).mean()
macro_f1 = f1_score(Y_test, rf_pred, average="macro")
micro_f1 = f1_score(Y_test, rf_pred, average="micro")

rf_overall_acc = overall_acc
rf_macro_f1 = macro_f1
rf_micro_f1 = micro_f1

print(f"Accuracy (overall): {overall_acc:.4f} ({overall_acc*100:.2f}%)")
print(f"F1 macro:           {macro_f1:.4f}")
print(f"F1 micro:           {micro_f1:.4f}")

per_f1 = []

for i, col in enumerate(LABEL_COLS):
    f1 = f1_score(Y_test.iloc[:, i], rf_pred[:, i])
    per_f1.append([col, f1])

print("\nPer-nutrient F1:")
display(pd.DataFrame(per_f1, columns=["nutrient", "f1"]))

rf_per_f1 = pd.DataFrame(
    per_f1,
    columns=["nutrient", "f1"]
)

rf_scores = {
    "accuracy": rf_overall_acc,
    "f1_macro": rf_macro_f1,
    "f1_micro": rf_micro_f1,
    "per_f1": rf_per_f1
}
```

```python
## Step 7: Train XGBoost

import time
from xgboost import XGBClassifier
from sklearn.multioutput import MultiOutputClassifier

print("Training XGBoost ...")

start = time.time()

xgb_model = MultiOutputClassifier(
    XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        tree_method="hist"
    )
)

xgb_model.fit(X_train_scaled, Y_train)

print(f"XGBoost done in {time.time()-start:.1f}s")

xgb_pred = xgb_model.predict(X_test_scaled)
```

```python
## Step 8: Evaluate XGBoost

from sklearn.metrics import f1_score
import pandas as pd

print("\n" + "="*70)
print("SCORES — XGBoost")
print("="*70)

# Element-wise accuracy (same as mentor)
overall_acc = (Y_test.values == xgb_pred).mean()

macro_f1 = f1_score(
    Y_test,
    xgb_pred,
    average="macro"
)

micro_f1 = f1_score(
    Y_test,
    xgb_pred,
    average="micro"
)

xgb_overall_acc = overall_acc
xgb_macro_f1 = macro_f1
xgb_micro_f1 = micro_f1
xgb_per_f1 = pd.DataFrame(per_f1, columns=["nutrient", "f1"])

print(f"Accuracy (overall): {overall_acc:.4f} ({overall_acc*100:.2f}%)")
print(f"F1 macro:           {macro_f1:.4f}")
print(f"F1 micro:           {micro_f1:.4f}")

per_f1 = []

for i, col in enumerate(LABEL_COLS):
    per_f1.append([
        col,
        f1_score(
            Y_test.iloc[:, i],
            xgb_pred[:, i]
        )
    ])

per_f1 = pd.DataFrame(
    per_f1,
    columns=["nutrient", "f1"]
)

print("\nPer-nutrient F1:")
display(per_f1)

xgb_per_f1 = pd.DataFrame(
    per_f1,
    columns=["nutrient", "f1"]
)

xgb_scores = {
    "accuracy": xgb_overall_acc,
    "f1_macro": xgb_macro_f1,
    "f1_micro": xgb_micro_f1,
    "per_f1": xgb_per_f1
}
```
