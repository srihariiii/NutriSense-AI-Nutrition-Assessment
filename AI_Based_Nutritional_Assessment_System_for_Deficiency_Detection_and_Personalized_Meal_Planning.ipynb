# Task 3

```python
## Day-2 Task: Select Required Columns and Rename Columns

# Required columns from each dataset
DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]

# Select only required columns
demo = demo[DEMO_CDC_COLS]
dr1tot = dr1tot[DIET_CDC_COLS]
cbc = cbc[CBC_CDC_COLS]
vid = vid[VID_CDC_COLS]
fertin = fertin[FER_CDC_COLS]

# Rename dictionary
NHANES_RENAME = {
    "SEQN": "person_id",
    "RIAGENDR": "gender",
    "RIDAGEYR": "age_years",
    "DR1TKCAL": "calories_kcal",
    "DR1TPROT": "protein_g",
    "DR1TIRON": "iron_mg",
    "DR1TCALC": "calcium_mg",
    "DR1TVD": "vitamin_d_mcg",
    "DR1TVB12": "vitamin_b12_mcg",
    "DR1TFOLA": "folate_mcg",
    "DR1TVARA": "vitamin_a_mcg",
    "DR1TVC": "vitamin_c_mg",
    "DR1TMAGN": "magnesium_mg",
    "DR1TZINC": "zinc_mg",
    "LBXHGB": "hemoglobin_g_dl",
    "LBXVIDMS": "serum_vitamin_d",
    "LBXFER": "ferritin",
}

# Rename columns
demo.rename(columns=NHANES_RENAME, inplace=True)
dr1tot.rename(columns=NHANES_RENAME, inplace=True)
cbc.rename(columns=NHANES_RENAME, inplace=True)
vid.rename(columns=NHANES_RENAME, inplace=True)
fertin.rename(columns=NHANES_RENAME, inplace=True)

# Display final output
print("DEMO Dataset")
display(demo.head())

print("DR1TOT Dataset")
display(dr1tot.head())

print("CBC Dataset")
display(cbc.head())

print("VID Dataset")
display(vid.head())

print("FERTIN Dataset")
display(fertin.head())
```

# Task 4

```python
## Step 1: Merge NHANES Datasets

df = demo.merge(dr1tot, on="person_id", how="inner")
df = df.merge(cbc, on="person_id", how="left")
df = df.merge(vid, on="person_id", how="left")
df = df.merge(fertin, on="person_id", how="left")

```

```python
## Step 2: Keep Adults (18–65 Years)

df = df[(df["age_years"] >= 18) & (df["age_years"] <= 65)].copy()
```

```python
## Step 3: Preview the Merged Dataset

print("Merged Dataset (Adults 18–65)")
print("Shape:", df.shape)

display(df.head())
```

```python
## Step 3: Remove Rows with Missing Required Columns

COLUMNS_NEEDED = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

df = df.dropna(subset=COLUMNS_NEEDED).copy()

## Step 4: Calculate Iron Label

df["label_iron"] = (
    (
        ((df["gender"] == 1) & (df["hemoglobin_g_dl"] < 13)) |
        ((df["gender"] == 2) & (df["hemoglobin_g_dl"] < 12)) |
        (df["ferritin"] < 30)
    )
    |
    (df["iron_mg"] < 8)
).astype(int)

## Step 5: Calculate Vitamin D Label

df["label_vitamin_d"] = (
    (df["serum_vitamin_d"] < 50) |
    (df["vitamin_d_mcg"] < 15)
).astype(int)

## Step 6: Calculate Vitamin B12 Label

df["label_vitamin_b12"] = (
    df["vitamin_b12_mcg"] < 2.4
).astype(int)

## Step 7: Calculate Calcium Label

df["label_calcium"] = (
    df["calcium_mg"] < 1000
).astype(int)

## Step 8: Calculate Folate Label

df["label_folate"] = (
    df["folate_mcg"] < 400
).astype(int)

## Step 9: Calculate Vitamin A Label

df["label_vitamin_a"] = (
    df["vitamin_a_mcg"] < 700
).astype(int)

## Step 10: Calculate Vitamin C Label

df["label_vitamin_c"] = (
    df["vitamin_c_mg"] < 75
).astype(int)

## Step 11: Calculate Magnesium Label

df["label_magnesium"] = (
    df["magnesium_mg"] < 320
).astype(int)

## Step 12: Calculate Zinc Label

df["label_zinc"] = (
    df["zinc_mg"] < 8
).astype(int)

## Step 13: Calculate Protein Label

df["label_protein"] = (
    df["protein_g"] < 50
).astype(int)
```

```python
## Step 14: Label Columns

LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]
```

```python
## Step 15: Preview Final Dataset

display(df.head(10))
print("Rows, Columns :", df.shape)
```

```python
## Label Balance

print("Label balance (0 vs 1):")

for col in LABEL_COLS:
    counts = df[col].value_counts().to_dict()
    positive = df[col].mean() * 100
    print(f"{col:22} → {counts} | positive={positive:.1f}%")
```
